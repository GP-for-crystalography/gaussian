```python
import h5py, posixpath

SRC = "masterKTN_014_01.h5"
DST = "masterKTN_014_01_slim.h5"
KEEP = {"mLPNH", "mLPSH"}      # duże zbiory, które model czyta
SMALL = 1_000_000              # bajtów: metadane kopiujemy zawsze

with h5py.File(SRC, "r") as f, h5py.File(DST, "w") as g:
    def visit(name, obj):
        if not isinstance(obj, h5py.Dataset):
            return
        big = obj.size * obj.dtype.itemsize > SMALL
        if big and posixpath.basename(name) not in KEEP:
            print("pomijam:", name, obj.shape)
            return
        g.require_group(posixpath.dirname(name))
        kw = dict(compression="gzip", compression_opts=4) if big else {}
        g.create_dataset(name, data=obj[()], **kw)
        print("kopiuję:", name, obj.shape, obj.dtype)
    f.visititems(visit)
```
